```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt
```

```python
dataset_path = Path(r"D:\good_projects\visionguard\mvtec_anomaly_detection")

category = "bottle"

train_good_path = dataset_path / category / "train" / "good"
test_path = dataset_path / category / "test"
ground_truth_path = dataset_path / category / "ground_truth"
```

```python
test_categories = [item.name for item in test_path.iterdir() if item.is_dir()]

print(test_categories)
```

```python
for category_name in test_categories:
    category_dir = test_path / category_name
    images = [item for item in category_dir.iterdir() if item.is_file()]
    print(category_name, ":", len(images))
```

```python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))

for ax, category_name in zip(axes, test_categories):
    category_dir = test_path / category_name
    image_path = next(category_dir.iterdir())

    image = Image.open(image_path)

    ax.imshow(image)
    ax.set_title(category_name)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
for category_name in test_categories:
    category_dir = test_path / category_name
    image_path = next(category_dir.iterdir())

    with Image.open(image_path) as image:
        print(category_name, image.size)
```

```python
fig, axes = plt.subplots(4, 3, figsize=(10, 12))

for row, category_name in enumerate(test_categories):
    category_dir = test_path / category_name
    image_paths = [
        item for item in category_dir.iterdir()
        if item.is_file()
    ][:3]

    for col, image_path in enumerate(image_paths):
        image = Image.open(image_path)

        axes[row, col].imshow(image)
        axes[row, col].set_title(category_name)
        axes[row, col].axis("off")

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

image_paths = [
    item for item in train_good_path.iterdir()
    if item.is_file()
][:3]

for ax, image_path in zip(axes, image_paths):
    image = Image.open(image_path)

    ax.imshow(image)
    ax.axis("off")

plt.tight_layout()
plt.show()
```
